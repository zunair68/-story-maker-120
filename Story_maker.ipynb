```python
%%writefile app.py
import streamlit as st
import time
import random

st.set_page_config(page_title="AI Story Maker", page_icon="📝", layout="wide")

# Custom dark-theme formatting
st.markdown("""
<style>
    .reportview-container { background: #121212; }
    .story-box { background-color: #1e1e24; border-left: 5px solid #ff4b4b; padding: 20px; border-radius: 5px; font-size: 18px; line-height: 1.6; font-family: 'Georgia', serif; color: #f0f2f6; }
    .stButton>button { width: 100%; border-radius: 20px; background-color: #ff4b4b; color: white; font-weight: bold; }
    .stButton>button:hover { background-color: #ff3333; color: white; border: 1px solid #ff3333; }
</style>
""", unsafe_allow_html=True)

st.title("📝 The AI Story Maker")
st.subheader("Craft unique, unexpected stories by twisting narrative metrics.")

# Sidebar Configuration Control Panels
st.sidebar.header("🎬 Narrative Engine Setup")
genre = st.sidebar.selectbox("Choose a Genre", ["Cyberpunk Sci-Fi", "Dark Cosmic Horror", "Fantasy Adventure", "Noir Detective Mystery"])
character_role = st.sidebar.text_input("Protagonist Name/Role", value="Neo-Netrunner Nova")
plot_twist_level = st.sidebar.slider("Plot Twist Intensity", 0, 100, 50)

# Narrative fragments database
story_fragments = {
    "Cyberpunk Sci-Fi": {
        "intro": "The neon rain slicked the micro-circuits of {hero}'s cybernetic arm. The megacity hummed, chewing up souls and spitting out data.",
        "twists": [
            "Suddenly, a terminal pinged. The digital identity file belonged to someone who died fifty years ago—themselves.",
            "A rogue corporate AI hijacked their visual neural lace, showing them a live feed of their own room from behind.",
            "The synthetic street drug they took didn't boost focus; it was uploading their memories directly onto the dark grid."
        ]
    },
    "Dark Cosmic Horror": {
        "intro": "The dusty old grandfather clock in {hero}'s study began ticking backward. The ancient leather-bound journal hummed with a soft, green luminescent glow.",
        "twists": [
            "When they looked into the mirror, their reflection didn't look back; it stood perfectly still, smiling.",
            "They realized the continuous scratching inside the stone walls wasn't rats—it was matching their exact heartbeat.",
            "The constellation map in the library slowly shifted, forming an eye that blinked straight down at the house."
        ]
    },
    "Fantasy Adventure": {
        "intro": "Deep inside the Whispering Caves, the rusted broadsword carried by {hero} burst into a magnificent, crackling purple fire.",
        "twists": [
            "The ancient dragon didn't breathe fire; it spoke in a clear, mechanical voice, asking for a systems update.",
            "The magic map didn't point toward gold; it was tracking the movements of the kingdom's beloved King, who was currently hiding behind them.",
            "As they grabbed the sapphire amulet, their companion turned into a swirling cloud of sand, muttering a forgotten curse."
        ]
    },
    "Noir Detective Mystery": {
        "intro": "The ceiling fan chopped through the heavy cigar smoke in {hero}'s office. The clock struck midnight when the brass door handle slowly turned.",
        "twists": [
            "The mysterious client dropped a folder containing photographs of the crime scene—taken three hours into the future.",
            "The bullet recovered from the wall didn't match any gun in the city; it was carved entirely out of absolute ice.",
            "The lead suspect walked in, wearing the exact same trench coat, carrying a badge with {hero}'s own identification number."
        ]
    }
}

st.write("---")

if st.button("🚀 GENERATE STORY CORE"):
    status_msg = st.empty()
    progress_bar = st.progress(0)

    layers = ["Structuring narrative bones...", "Fleshing out emotional anchors...", "Injecting chaotic random variables..."]
    for idx, p in enumerate([33, 66, 100]):
        time.sleep(0.4)
        progress_bar.progress(p)
        status_msg.markdown(f"🖋️ *{layers[idx]}*")

    status_msg.success("✨ Story Matrix Synced Successfully!")

    data = story_fragments[genre]
    base_story = data["intro"].format(hero=character_role)

    if plot_twist_level < 30:
        twist_text = "The path ahead remained quiet, but an unsettling shadow stretched out behind them, promising that the real trial was yet to come."
    elif plot_twist_level < 75:
        twist_text = random.choice(data["twists"])
    else:
        twist_text = f"**{random.choice(data['twists'])}** Instantly, time fractured. Reality peeled back like burnt paper, leaving {character_role} stranded inside a dimensional loop."

    full_narrative = f"{base_story} \n\n {twist_text}"

    st.markdown("### 📖 The Resulting Tale")
    st.markdown(f"<div class='story-box'>{full_narrative}</div>", unsafe_allow_html=True)
```
